# D-Fire — Phase 7: Demo Inference

Loads a trained detector and runs it on sample images, showing annotated
detections and the operational alert state. Compares operating modes
(high_recall / balanced / high_precision).

```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd()
while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / "src").is_dir():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT))

WEIGHTS = str(REPO_ROOT / "outputs" / "weights" / "yolo_main" / "weights" / "best.pt")

from src.utils import resolve_data_root
from src.inference.detector import WildfireDetector
from src.visualization.plot_predictions import _read_rgb
from src.data.dfire import iter_split

data_root = resolve_data_root(None)
sample_images = [p for p, _ in iter_split(data_root / "test" / "images")][:6]
print("Samples:", [p.name for p in sample_images])
```

```python
import matplotlib.pyplot as plt

detector = WildfireDetector.from_config(REPO_ROOT / "configs" / "inference.yaml", mode="balanced", weights=WEIGHTS)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, img_path in zip(axes.ravel(), sample_images):
    rgb = _read_rgb(img_path)
    dets = detector.predict(str(img_path))
    alert = detector.alert_state(dets)
    ax.imshow(detector.annotate(rgb, dets))
    ax.set_title(f"{img_path.name} -> {alert.value}", fontsize=9)
    ax.axis("off")
plt.tight_layout(); plt.show()
```
